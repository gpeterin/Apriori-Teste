```python
import pandas as pd
import numpy as np

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
df = pd.read_csv("./compras.csv", header=None)
df.sample(10)
```

```python
df.replace(np.nan, 0, inplace = True)
df.sample(10)
```

```python
transacoes = []

for index, row in df.iterrows():
    transacao_unica = row.values.tolist()
    transacao_unica = [item for item in transacao_unica if item != 0]
    transacoes.append(transacao_unica)

print(transacao_unica[0:10])
```

```python
enconder = TransactionEncoder()
matriz = enconder.fit(transacoes).transform(transacoes)

df_transacoes = pd.DataFrame(matriz, columns=enconder.columns_)
df_transacoes.head()
```

```python
itens_frequentes = apriori(df_transacoes, min_support = 0.01, use_colnames = True)
itens_frequentes.sort_values(by=['support'], ascending = False).head(10)
```

```python
regras = association_rules(itens_frequentes, metric="confidence", min_threshold=0.3)
regras.sort_values(by=['lift'], ascending = False).drop(['antecedent support', 'consequent support', 'leverage', 'conviction'], axis=1)
```
